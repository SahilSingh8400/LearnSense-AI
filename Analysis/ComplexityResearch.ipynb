```python
import pandas as pd

df = pd.DataFrame(ds)

df["length"] = df["src"].apply(len)

df.groupby("complexity")["length"].mean().sort_values()
```

```python
for c in sorted(set(ds["complexity"])):
    sample = next(x for x in ds if x["complexity"] == c)

    print("\n" + "="*80)
    print("CLASS:", c)
    print("="*80)

    print(sample["src"][:1500])
```

```python
import re

def count_loops(code):
    return len(re.findall(r'\b(for|while)\b', code))

df["loop_count"] = df["src"].apply(count_loops)

print(
    df.groupby("complexity")["loop_count"]
      .mean()
      .sort_values()
)
```
